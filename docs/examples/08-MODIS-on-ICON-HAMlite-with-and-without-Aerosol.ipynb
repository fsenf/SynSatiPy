# Run Synsat on ICON-HAMlite Data for MODIS with and without Aerosol

Here, we will 
- load ICON-HAMlite test data (limited area simulation, stored in separate files, from BioSmoke project)
- run Synsat on ICON for MODIS with without Aerosol and
- finally store the Synsats into a netcdf file

The workflow is similar to [04-Run-Synsat-on-ICON-Data-for-SEVIRI.ipynb](04-Run-Synsat-on-ICON-Data-for-SEVIRI.ipynb).


## Setup Env and Load Libraries 

```python
import os, sys, gc

os.environ["RTTOV_PYTHON_WRAPPER"] = "/work/bb1262/tools/rttov/rttov-v13.2/wrapper"
cores = 4  #  int( os.environ.get("SLURM_CPUS_ON_NODE") )

import xarray as xr
import pylab as plt

import warnings

warnings.filterwarnings("ignore")
```

```python
print(f"{cores} cores for calculations available")
```

```python
import synsatipy

from synsatipy.synsat import SynSat

import synsatipy.synsat_example_data as synsat_example_data
import synsatipy.utils.colormaps as colormaps
```

## Steps towards Synsat 

### Synsat Workflow without Aerosol 

No Aerosol is the default setting and nothing needs to be added.

```python
print("Step I: Initialize Synsat class")
print("-------------------------------")
s_no_aerosol = SynSat(synsat_instrument="MODIS")
print()


print("Step II: Load Example Data")
print("--------------------------")
iconname, geofile = synsat_example_data.get_example_data(
    "hamlite01", return_geofile=True
)
isel = {"ncells": slice(1000, None, 12)}
s_no_aerosol.load(
    iconname, isel=isel, profile_dimensions=["time", "ncells"], geofile=geofile
)
print()


print("Step III: Setup Run Options and Start Execution")
print("-----------------------------------------------")
s_no_aerosol._options.Nthreads = cores
s_no_aerosol._options.NprofsPerCall = 1000
s_no_aerosol.run(chunked=True)
```

```python
print("Step IV: Extract the Output")
print("---------------------------")
icon_synsat_na = s_no_aerosol.extract_output()

indat = s_no_aerosol.synsat.data_handler.input_data
icon_synsat_na["lon"] = indat.lon
icon_synsat_na["lat"] = indat.lat

del s_no_aerosol
print(icon_synsat_na)
print()


print("Step V: Store Data")
print("------------------")


outdir = "./Data"
if not os.path.isdir(outdir):
    os.makedirs(outdir)
outfile = f"{outdir}/modis_synsat_iconhamlite_no-aerosol.nc"

print(f"... store synsat at {outfile}")
icon_synsat_na.to_netcdf(outfile)
```

### Synsat Workflow with Aerosols 

Aerosol needs two main additions
1. synsat class needs to be initialized with `synsat_use_aerosols=True`
2. an `aerosol_config` needs to be provided to the data loader

```python
synsat_use_aerosols = True

aerosol_config = {
    "num_qca": {
        "radius": 8e-08,
        "density": 1993.0,
        "sigma": 1.59,
        "CAMS_OPAC_name": "BCAR",
        "target_name": "qca",
    },
    "num_qsu": {
        "radius": 1.4e-07,
        "density": 1859.0,
        "sigma": 1.59,
        "CAMS_OPAC_name": "SULP",
        "target_name": "qsu",
    },
    "num_qss": {
        "radius": 8.5e-07,
        "density": 2165.0,
        "sigma": 2.0,
        "CAMS_OPAC_name": "SSA2",
        "target_name": "qss",
    },
    "num_qdu": {
        "radius": 6e-07,
        "density": 2650.0,
        "sigma": 2.0,
        "CAMS_OPAC_name": "DUS2",
        "target_name": "qdu",
    },
}
```

```python
print("Step I: Initialize Synsat class")
print("-------------------------------")
s_with_aerosol = SynSat(
    synsat_instrument="MODIS", synsat_use_aerosols=synsat_use_aerosols
)
print()


print("Step IIa: Define Aerosol Config")
print("-------------------------------")

print("Step IIb: Load Example Data")
print("--------------------------")
iconname, geofile = synsat_example_data.get_example_data(
    "hamlite01", return_geofile=True
)
s_with_aerosol.load(
    iconname,
    isel=isel,
    profile_dimensions=["time", "ncells"],
    geofile=geofile,
    aerosol_config=aerosol_config,
)
print()


print("Step III: Setup Run Options and Start Execution")
print("-----------------------------------------------")
s_with_aerosol._options.Nthreads = cores
s_with_aerosol._options.NprofsPerCall = 1000
s_with_aerosol.run(chunked=True)
```

```python
print("Step IV: Extract the Output")
print("---------------------------")
icon_synsat_wa = s_with_aerosol.extract_output()

indat = s_with_aerosol.synsat.data_handler.input_data
icon_synsat_wa["lon"] = indat.lon
icon_synsat_wa["lat"] = indat.lat

del s_with_aerosol
print(icon_synsat_wa)
print()


print("Step V: Store Data")
print("------------------")


outdir = "./Data"
if not os.path.isdir(outdir):
    os.makedirs(outdir)
outfile = f"{outdir}/modis_synsat_iconhamlite_with-aerosol.nc"

print(f"... store synsat at {outfile}")
icon_synsat_wa.to_netcdf(outfile)
```

## Plotting

```python
x, y = icon_synsat_na["lon"], icon_synsat_na["lat"]
v1 = icon_synsat_na["bt110"].squeeze()
v2 = icon_synsat_wa["bt110"].squeeze()
```

```python
pkws = dict(
    cmap=colormaps.enhanced_colormap(vmin=210, vmed=240, vmax=300),
    vmin=210,
    vmax=300,
)
```

```python
fig, axs = plt.subplots(ncols=3, figsize=(18, 7))
plt.sca(axs[0])
plt.tripcolor(x, y, v1, **pkws)
plt.title("(a) Canadian Fires: Synsat without Aerosol")
plt.colorbar(orientation="horizontal", label="MODIS BT at 11 um")

plt.sca(axs[1])
plt.tripcolor(x, y, v2, **pkws)
plt.title("(b) Canadian Fires: Synsat with Aerosol")
plt.colorbar(orientation="horizontal", label="MODIS BT at 11 um")

plt.sca(axs[2])
plt.tripcolor(x, y, v2 - v1, vmin=-5, vmax=5, cmap=plt.cm.RdBu_r)
plt.title("(c) Canadian Fires: BT Difference (Aer - noAer)")
plt.colorbar(orientation="horizontal", label="MODIS BT Difference at 11 um")

plt.tight_layout()
```
